```python
###META_data


import json
import re
from bert_score import score
from tqdm import tqdm
# -----------------------
# CONFIG: INPUT/OUTPUT PATHS
# -----------------------
GT_PATH = /685/Australian_data/code/baseline_retrieved_prompts.json"               # Ground-truth JSON
GEN_PATH = /685/Australian_data/logs/llama1b/retrieved_baseline_positive_best/generated_answers.json"       # Generated answers JSON
OUTPUT_PATH = /685/Australian_data/logs/llama1b/retrieved_baseline_positive_best/bertscore_results.json"  # Output JSON

# -----------------------
# HELPER FUNCTION
# -----------------------
def extract_response_text(s):
    """Extract response_text from generated answer string."""
    if not isinstance(s, str):
        return ""
    match = re.search(r"response_text=['\"](.+?)['\"]", s, re.DOTALL)
    return match.group(1).strip() if match else s.strip()

# -----------------------
# LOAD JSON FILES
# -----------------------
with open(GT_PATH, "r", encoding="utf-8") as f:
    gt = json.load(f)

with open(GEN_PATH, "r", encoding="utf-8") as f:
    gen = json.load(f)

# -----------------------
# COLLECT REFERENCES AND CANDIDATES
# -----------------------
refs = []
cands = []
keys_used = []

for qid in tqdm(gt):
    if qid in gen:
        ref_answer = gt[qid]["answer"]
        gen_answer = extract_response_text(gen[qid]["generated_answer"])
        refs.append(ref_answer)
        cands.append(gen_answer)
        keys_used.append(qid)

print(f"Number of QA pairs: {len(cands)}")

# -----------------------
# COMPUTE BERTSCORE
# -----------------------
P, R, F1 = score(cands, refs, lang="en", model_type="roberta-large", batch_size=8, verbose=True)

# -----------------------
# SAVE RESULTS
# -----------------------
bert_results = {}
for i, qid in enumerate(keys_used):
    bert_results[qid] = {
        "precision": P[i].item(),
        "recall": R[i].item(),
        "f1": F1[i].item()
    }

with open(OUTPUT_PATH, "w", encoding="utf-8") as f:
    json.dump(bert_results, f, indent=2)

# Print average F1
print(f"\nAverage BERTScore F1: {F1.mean().item():.4f}")
```
