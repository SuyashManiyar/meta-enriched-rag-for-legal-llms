```python
import pandas as pd
import re
import logging
from datetime import datetime

# -------------------------------
# 1. USER CONFIG
# -------------------------------
# e.g. "../../Dataset/Test/data.tsv"
log_filename = "/home/smaniyar_umass_edu/BioNLP_Ontology/other/nlp/meta-enriched-rag-for-legal-llms/privacyQA/test_results/finetned_llama3b/logfile.log" # e.g. "/home/user/privacyQA/logs/run1.log"
out_csv = "/home/smaniyar_umass_edu/BioNLP_Ontology/other/nlp/meta-enriched-rag-for-legal-llms/privacyQA/test_results/finetned_llama3b/predictions.csv"

# -------------------------------
# 2. SETUP LOGGING
# -------------------------------
logging.basicConfig(
    filename=log_filename,
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)
logging.info("Starting classification run using Llama Factory chat model...")


# -------------------------------
def make_prompt(row):
    return (
        "You are a legal expert, classifier that determines if a segment of text "
        "from a document is relevant to a user query.\n\n"
        f'Query: "{row["Query"]}"\n'
        f'Segment: "{row["Segment"]}"\n\n'
        "Answer whether the segment is RELEVANT or IRRELEVANT to the query. "
        'Answer only with "RELEVANT" or "IRRELEVANT".'
    )

# -------------------------------
# 6. RUN CHAT INFERENCE + LOGGING
# -------------------------------
results = []

df=pd.read_csv('/home/smaniyar_umass_edu/BioNLP_Ontology/other/nlp/meta-enriched-rag-for-legal-llms/privacyQA/data/stratified_test_sample_1000.csv')
for idx, row in df.iterrows():
    queryid = row["DocID"]+"_"+row["QueryID"]+"_"+row["SentID"]
    prompt = make_prompt(row)
    actual_label=row['Any_Relevant']
    messages = [{"role": "user", "content": prompt}]

    try:
        response = chat_model.chat(messages)
        raw_output = response["response"] if isinstance(response, dict) and "response" in response else str(response)
    except Exception as e:
        logging.error(f"[{queryid}] Error during chat: {e}")
        raw_output = "ERROR"

    # Extract RELEVANT or IRRELEVANT
    match = re.search(r"\b(RELEVANT|IRRELEVANT)\b", raw_output, re.IGNORECASE)
    extracted = match.group(1).upper() if match else "NONE"

    # Log full details
    logging.info(
        f"[{queryid}] PROMPT:\n{prompt}\n"
        f"RAW OUTPUT: {raw_output}\nEXTRACTED: {extracted}\n{'-'*80}"
    )

    # Append for CSV
    results.append({
        "QueryID": queryid,
        "RawOutput": raw_output,
        "ExtractedOutput": extracted,
        "GT":actual_label

    })

# -------------------------------
# 7. SAVE RESULTS TO CSV
# -------------------------------
pd.DataFrame(results).to_csv(out_csv, index=False)
logging.info(f"Saved predictions to {out_csv}")
logging.info("Run complete.")
```
